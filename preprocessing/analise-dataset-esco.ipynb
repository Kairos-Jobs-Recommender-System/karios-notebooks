```python
# Classificação baseada na estrutura ESCO para agrupar competências similares.

skill_groups_df = pd.read_csv("esco-dataset/skillGroups_pt.csv")
skill_groups_df.head()
```

```python
# Get all skills (from skills_pt) where is connected with skill_groups_df[0]

# 1. Carregar as relações de hierarquia
df_broader_skills = pd.read_csv("esco-dataset/broaderRelationsSkillPillar_pt.csv")

# 2. Obter o grupo alvo (primeiro item do dataframe de grupos)
target_group = skill_groups_df.iloc[0]
target_uri = target_group['conceptUri']
target_code = str(target_group['code'])

# 3. Identificar todos os grupos descendentes (incluindo o próprio)
# Na ESCO (ISCED-F), a hierarquia é refletida no prefixo do código
descendant_groups = skill_groups_df[skill_groups_df['code'].astype(str).str.startswith(target_code)]
descendant_uris = set(descendant_groups['conceptUri'])

# 4. Encontrar skills que tenham qualquer um desses grupos como "broader"
# Filtramos a tabela de relações onde o broaderUri está no nosso conjunto de URIs descendentes
connected_relations = df_broader_skills[df_broader_skills['broaderUri'].isin(descendant_uris)]

# 5. Filtrar df_skills para obter os detalhes das skills encontradas
skills_connected = df_skills[df_skills['conceptUri'].isin(connected_relations['conceptUri'])]

# Exibir resultado
print(f"Grupo alvo: {target_group['preferredLabel']} ({target_code})")
print(f"Total de subgrupos encontrados: {len(descendant_groups)}")
print(f"Total de skills conectadas: {len(skills_connected)}")

skills_connected.head()
```

# Campos Comuns Importantes

Em quase todos os arquivos, você encontrará:

- `conceptUri`: O link permanente (ex: http://data.europa.eu/esco/occupation/756...). É a chave primária para fazer JOINS entre os arquivos.

- `preferredLabel`: O nome oficial do termo em português.

- `altLabels`: Uma lista de termos alternativos separados por quebra de linha (útil para sistemas de busca/SEO).
